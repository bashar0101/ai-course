## بالكود

</div>

```python
import numpy as np

a = np.array([2, 1])
b = np.array([1, 3])

print("a + b    =", a + b)
print("2 * a    =", 2 * a)
print("|a|      =", round(np.linalg.norm(a), 3))
print("distance =", round(np.linalg.norm(a - b), 3))
print("a . b    =", np.dot(a, b), "=", a @ b)
```

<div dir="rtl">

### دالة التشابه بجيب التمام

</div>

```python
def cosine(u, v):
    return (u @ v) / (np.linalg.norm(u) * np.linalg.norm(v))

print(round(cosine(a, b), 3))
print(round(cosine(a, -a), 3))
print(round(cosine(np.array([1, 0]), np.array([0, 1])), 3))
```

### متجهات كلمات: لمحة عمّا سيأتي

</div>

```python
words = {
    "king":  np.array([0.9, 0.8, 0.1]),
    "queen": np.array([0.8, 0.9, 0.1]),
    "apple": np.array([0.1, 0.2, 0.9]),
}
for w1, w2 in [("king", "queen"), ("king", "apple"), ("queen", "apple")]:
    print(f"{w1:6} ~ {w2:6}: {cosine(words[w1], words[w2]):.3f}")
```

### مشكلة الوحدات، بالكود

```python
flats = np.array([[100, 3],
                  [102, 1],
                  [ 80, 3]])
d_ab = np.linalg.norm(flats[0] - flats[1])
d_ac = np.linalg.norm(flats[0] - flats[2])
print(f"flat 1 to flat 2: {d_ab:.2f}")
print(f"flat 1 to flat 3: {d_ac:.2f}")
```
