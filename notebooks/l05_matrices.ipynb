```python
import numpy as np

X = np.array([[1, 2],
              [3, 1],
              [0, 4]])
w = np.array([2, 1])
b = 1

y_hat = X @ w + b
print("X shape:", X.shape, " w shape:", w.shape)
print("predictions:", y_hat)
```

```python
A = np.array([[1, 0, 2],
              [-1, 3, 1]])
B = np.array([[3, 1],
              [2, 1],
              [1, 0]])

print(A @ B)
print("A @ B shape:", (A @ B).shape)
print("B @ A shape:", (B @ A).shape)
print(A.T)
```

```python
rng = np.random.default_rng(0)
X4 = rng.normal(size=(4, 3))    # 4 أمثلة، 3 ميزات
W = rng.normal(size=(3, 2))     # 3 مدخلات، 2 عصبون
b2 = np.array([0.1, -0.2])

H = X4 @ W + b2
print("H shape:", H.shape)
print(H.round(3))
```
