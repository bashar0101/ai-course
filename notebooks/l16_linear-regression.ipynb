### نصنع البيانات

</div>

```python
import numpy as np

rng = np.random.default_rng(42)
n = 40
years = rng.uniform(0, 10, size=n)
salary = 30 + 5 * years + rng.normal(0, 4, size=n)   # القاعدة الحقيقية + ضجيج

print("first 5 years :", years[:5].round(1))
print("first 5 salary:", salary[:5].round(1))
```

<div dir="rtl">

### الحل المباشر

</div>

```python
def fit_closed_form(x, y):
    w = np.sum((x - x.mean()) * (y - y.mean())) / np.sum((x - x.mean()) ** 2)
    b = y.mean() - w * x.mean()
    return w, b

w_cf, b_cf = fit_closed_form(years, salary)
print(f"closed form: w = {w_cf:.3f}, b = {b_cf:.3f}")
```

<div dir="rtl">

### من الصفر، بالنزول بالتدرّج

</div>

```python
class LinearRegressionGD:
    def __init__(self, lr=0.01, epochs=5000):
        self.lr = lr
        self.epochs = epochs

    def fit(self, x, y):
        self.w, self.b = 0.0, 0.0
        self.history = []
        for _ in range(self.epochs):
            err = self.predict(x) - y
            self.w -= self.lr * 2 * np.mean(err * x)
            self.b -= self.lr * 2 * np.mean(err)
            self.history.append(np.mean(err ** 2))
        return self

    def predict(self, x):
        return self.w * x + self.b

model = LinearRegressionGD(lr=0.01, epochs=5000).fit(years, salary)
print(f"gradient descent: w = {model.w:.3f}, b = {model.b:.3f}")
print(f"final MSE: {model.history[-1]:.3f}")
```

<div dir="rtl">

### بمكتبة سايكت ليرن

</div>

```python
from sklearn.linear_model import LinearRegression

X = years.reshape(-1, 1)          # المكتبة تنتظر جدولاً: صف لكل مثال
lr_model = LinearRegression().fit(X, salary)
print(f"sklearn: w = {lr_model.coef_[0]:.3f}, b = {lr_model.intercept_:.3f}")
print("prediction for 7 years:", lr_model.predict([[7]]).round(2))
```

<div dir="rtl">

### نرسم الخط والبواقي

</div>

```python
import matplotlib.pyplot as plt

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(9, 3.4))

xx = np.linspace(0, 10, 50)
ax1.scatter(years, salary, s=25, color="C1", label="data")
ax1.plot(xx, w_cf * xx + b_cf, lw=2.5, label="fitted line")
ax1.set_xlabel("years of experience"); ax1.set_ylabel("salary")
ax1.legend(); ax1.grid(True)

residuals = salary - (w_cf * years + b_cf)
ax2.scatter(years, residuals, s=25)
ax2.axhline(0, color="gray", lw=1.5)
ax2.set_xlabel("years of experience"); ax2.set_ylabel("residual")
ax2.grid(True)
plt.tight_layout()
plt.show()
```
