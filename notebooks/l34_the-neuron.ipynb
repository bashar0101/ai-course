## دوال التفعيل المشهورة

</div>

```python
import numpy as np
import matplotlib.pyplot as plt

z = np.linspace(-4, 4, 400)
fig, axes = plt.subplots(1, 4, figsize=(11, 2.6), sharey=True)
funcs = [("step", (z >= 0).astype(float)), ("sigmoid", 1 / (1 + np.exp(-z))),
         ("tanh", np.tanh(z)), ("ReLU", np.maximum(0, z))]
for ax, (name, v) in zip(axes, funcs):
    ax.plot(z, v, lw=2.5)
    ax.axhline(0, color="gray", lw=0.8); ax.axvline(0, color="gray", lw=0.8)
    ax.set_title(name); ax.set_ylim(-1.3, 2.5); ax.grid(True)
plt.tight_layout()
plt.show()
```

### عصبون في نمباي

</div>

```python
import numpy as np

def relu(z):
    return np.maximum(0, z)

def sigmoid(z):
    return 1 / (1 + np.exp(-z))

def step(z):
    return (z >= 0).astype(float)

def neuron(x, w, b, activation):
    z = x @ w + b
    return activation(z)

x = np.array([2.0, 1.0])
w = np.array([0.5, -1.0])
for f in [relu, sigmoid, np.tanh, step]:
    print(f"{f.__name__:8}: {neuron(x, w, 0.2, f):.3f}")
```

### البوابات، لكل المدخلات دفعة واحدة

</div>

```python
X = np.array([[0, 0], [0, 1], [1, 0], [1, 1]])
w = np.array([1.0, 1.0])
print("AND:", neuron(X, w, -1.5, step))
print("OR :", neuron(X, w, -0.5, step))
```

### محاولة يائسة: "إما أو" بعصبون واحد

</div>

```python
rng = np.random.default_rng(0)
target = np.array([0, 1, 1, 0])
best = 0
for _ in range(100_000):
    w = rng.uniform(-5, 5, 2)
    b = rng.uniform(-5, 5)
    correct = np.sum(neuron(X, w, b, step) == target)
    best = max(best, correct)
print("best result for XOR with one neuron:", best, "of 4")
```
