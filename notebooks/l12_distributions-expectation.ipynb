### نسحب عينات من توزيع

</div>

```python
import numpy as np

rng = np.random.default_rng(0)

rolls = rng.integers(1, 7, size=10)
print("10 dice rolls:", rolls)

words = ["al-khair", "al-noor", "al-full"]
picks = rng.choice(words, size=10, p=[0.7, 0.2, 0.1])
print("10 next words:", picks)
```

### قانون الأعداد الكبيرة

</div>

```python
import matplotlib.pyplot as plt

rolls = rng.integers(1, 7, size=10_000)
running_mean = np.cumsum(rolls) / np.arange(1, len(rolls) + 1)

plt.figure(figsize=(6, 3.4))
plt.plot(running_mean, lw=1.8)
plt.axhline(3.5, color="C1", ls="--", lw=2, label="expected value 3.5")
plt.xscale("log")
plt.xlabel("number of rolls (log scale)"); plt.ylabel("average so far")
plt.legend(); plt.grid(True)
plt.show()

print("average of 10 rolls    :", running_mean[9])
print("average of 10000 rolls :", round(running_mean[-1], 3))
```

<div dir="rtl">

### الجرس يظهر من النرد

</div>

```python
means = rng.integers(1, 7, size=(100_000, 10)).mean(axis=1)

plt.figure(figsize=(6, 3.4))
plt.hist(means, bins=np.arange(1, 6.05, 0.1) - 0.05, color="C0", alpha=0.8)
plt.xlabel("average of 10 dice"); plt.ylabel("count")
plt.show()

print("mean of the averages:", round(means.mean(), 3))
print("std of the averages :", round(means.std(), 3))
```

### نتحقق من قاعدة 68 و 95

</div>

```python
heights = rng.normal(loc=175, scale=7, size=100_000)
within_1 = np.mean(np.abs(heights - 175) < 7)
within_2 = np.mean(np.abs(heights - 175) < 14)
print(f"within 1 std: {within_1:.3f}   within 2 std: {within_2:.3f}")
```
