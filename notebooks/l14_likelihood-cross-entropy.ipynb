### منحنى الأرجحية لقطعة النقود

</div>

```python
import numpy as np
import matplotlib.pyplot as plt

p = np.linspace(0.01, 0.99, 300)
likelihood = p**3 * (1 - p)

plt.figure(figsize=(6, 3.2))
plt.plot(p, likelihood, lw=2.5)
plt.axvline(0.75, color="C1", ls="--", lw=2, label="best p = 0.75")
plt.xlabel("p = probability of heads"); plt.ylabel("likelihood of H,H,T,H")
plt.legend(); plt.grid(True)
plt.show()

print("best p on the grid:", round(p[np.argmax(likelihood)], 3))
```

### الإنتروبيا المتقاطعة من الصفر

</div>

```python
def cross_entropy(probs, correct):
    """probs: شكل (n, k)، احتمال كل فئة. correct: رقم الفئة الصحيحة لكل مثال."""
    p_correct = probs[np.arange(len(correct)), correct]
    return -np.mean(np.log(p_correct))

probs = np.array([[0.9, 0.05, 0.05],     # المثال 1: الصحيحة 0
                  [0.3, 0.5,  0.2 ],     # المثال 2: الصحيحة 1
                  [0.1, 0.1,  0.8 ]])    # المثال 3: الصحيحة 0
correct = np.array([0, 1, 0])

print("cross-entropy:", round(cross_entropy(probs, correct), 3))

from sklearn.metrics import log_loss
print("sklearn log_loss:", round(log_loss(correct, probs, labels=[0, 1, 2]), 3))
print("perplexity:", round(np.exp(cross_entropy(probs, correct)), 2))
```

### الإنتروبيا

</div>

```python
def entropy(p):
    p = np.asarray(p, dtype=float)
    p = p[p > 0]                 # صفر في لوغاريتم الصفر يساوي صفراً
    return -np.sum(p * np.log(p)) + 0.0   # + 0.0 يحوّل سالب الصفر إلى صفر

for dist in [[1, 0], [0.9, 0.1], [0.5, 0.5], [0.1] * 10]:
    print(f"{str(dist[:3]):18} entropy = {entropy(dist):.3f}")
```
