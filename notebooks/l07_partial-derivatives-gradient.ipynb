## نرى الأرض من فوق

</div>

```python
import numpy as np
import matplotlib.pyplot as plt

def f(w, b):
    return (w - 2) ** 2 + 2 * (b + 1) ** 2

W, B = np.meshgrid(np.linspace(-1, 4, 200), np.linspace(-3, 1.5, 200))
plt.figure(figsize=(6, 4.2))
cs = plt.contour(W, B, f(W, B), levels=[0.25, 1, 2, 4, 6, 9, 13], cmap="Blues")
plt.clabel(cs, fontsize=8)
plt.annotate("", xy=(0.8, -0.8), xytext=(0, 0),
             arrowprops=dict(arrowstyle="->", color="C1", lw=2.5))
plt.annotate("", xy=(-0.5, 0.5), xytext=(0, 0),
             arrowprops=dict(arrowstyle="->", color="C3", lw=2, ls="--"))
plt.scatter([0], [0], color="C1", zorder=3)
plt.scatter([2], [-1], marker="*", s=200, color="C2", zorder=3)
plt.text(0.1, 0.35, "gradient (uphill)", color="C3")
plt.text(0.9, -0.75, "minus gradient (downhill)", color="C1")
plt.text(2.1, -1.3, "minimum", color="C2")
plt.xlabel("w"); plt.ylabel("b")
plt.gca().set_aspect("equal")
plt.show()
```

### التدرّج بالقواعد، وبالحساب العددي

</div>

```python
def f(w, b):
    return (w - 2) ** 2 + 2 * (b + 1) ** 2

def grad(w, b):                      # من القواعد
    return (2 * w - 4, 4 * b + 4)

def numeric_grad(func, w, b, h=1e-5):
    dw = (func(w + h, b) - func(w - h, b)) / (2 * h)   # نحرك w فقط
    db = (func(w, b + h) - func(w, b - h)) / (2 * h)   # نحرك b فقط
    return (dw, db)

for point in [(0, 0), (1, 1), (2, -1)]:
    rule = grad(*point)
    num = numeric_grad(f, *point)
    print(point, "rule:", rule, " numeric:", tuple(round(v, 4) + 0.0 for v in num))
```

```python
w, b = 0.0, 0.0
for step in range(6):
    gw, gb = grad(w, b)
    print(f"step {step}: w = {w:6.3f}  b = {b:6.3f}  f = {f(w, b):.4f}")
    w, b = w - 0.1 * gw, b - 0.1 * gb
```
