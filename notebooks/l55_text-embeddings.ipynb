### البحث بالكلمات

</div>

```python
import numpy as np
from sklearn.feature_extraction.text import TfidfVectorizer

def evaluate(score_fn, questions, ks=(1, 3)):
    ranks = []
    for question, gold in questions:
        order = np.argsort(-score_fn(question), kind="stable")       # best page first
        ranks.append(int(np.where(order == index[gold])[0][0]) + 1)   # position of the right page
    ranks = np.array(ranks)
    return {f"recall@{k}": round(float((ranks <= k).mean()), 2) for k in ks}

word_vec = TfidfVectorizer()                      # one feature per word
word_docs = word_vec.fit_transform(texts)

def word_scores(question):
    return (word_vec.transform([question]) @ word_docs.T).toarray()[0]

q = "طلب مني الطبيب رنيناً مغناطيسياً"
print("known words in the question:", [w for w in word_vec.build_analyzer()(q) if w in word_vec.vocabulary_])
print("best score:", word_scores(q).max())
print("words:", evaluate(word_scores, test_q))
```

### قطع الحروف بدل الكلمات

```python
char_vec = TfidfVectorizer(analyzer="char_wb", ngram_range=(2, 4))   # pieces of 2 to 4 letters, inside words
char_docs = char_vec.fit_transform(texts)

def char_scores(question):
    return (char_vec.transform([question]) @ char_docs.T).toarray()[0]

shared = set(char_vec.build_analyzer()("رنيناً")) & set(char_vec.build_analyzer()("الرنين"))
print("pieces shared by the two words:", sorted(shared))
print("top page:", ids[char_scores(q).argmax()])
print("chars:", evaluate(char_scores, test_q))
```

### مرمّز صغير يتعلم من الأزواج

</div>

```python
import re
import zlib
import torch
import torch.nn as nn
import torch.nn.functional as F

BUCKETS = 2 ** 15

def pieces(text):
    """character pieces of every word, with word edges: <word> gives <w, wo, or, ..."""
    out = []
    for w in re.findall(r"\w+", text.lower()):
        w = f"<{w}>"
        for n in (2, 3, 4):
            out += [w[i:i + n] for i in range(len(w) - n + 1)]
    return out

class TinyEncoder(nn.Module):
    def __init__(self, dim=64):
        super().__init__()
        self.table = nn.EmbeddingBag(BUCKETS, dim, mode="mean")    # one vector per piece; a text is their mean
        nn.init.normal_(self.table.weight, std=0.1)

    def forward(self, batch):
        flat, starts = [], []
        for text in batch:
            starts.append(len(flat))
            flat += [zlib.crc32(p.encode()) % BUCKETS for p in pieces(text)]   # piece -> row number
        v = self.table(torch.tensor(flat), torch.tensor(starts))
        return F.normalize(v, dim=-1)                                # length 1: dot product = cosine

print(pieces("رنيناً")[:6], "...")
```

### التدريب

</div>

```python
torch.manual_seed(0)
rng = np.random.default_rng(0)
encoder = TinyEncoder()

def encoder_scores(question):
    with torch.no_grad():
        return (encoder([question]) @ encoder(texts).T)[0].numpy()

print("before training:", evaluate(encoder_scores, test_q))

questions_of = {}
for question, page in train_q:
    questions_of.setdefault(page, []).append(question)

opt = torch.optim.Adam(encoder.parameters(), lr=0.02)
tau, batch = 0.1, 16
for step in range(201):
    chosen = rng.choice(len(docs), size=batch, replace=False)      # 16 different pages
    qs = [rng.choice(questions_of[ids[j]]) for j in chosen]       # one question for each
    Q, D = encoder(qs), encoder([texts[j] for j in chosen])
    logits = Q @ D.T / tau                                         # 16 x 16 similarities
    loss = F.cross_entropy(logits, torch.arange(batch))            # the right page is on the diagonal
    opt.zero_grad()
    loss.backward()
    opt.step()
    if step % 50 == 0:
        print(f"step {step:3}  loss {loss.item():.3f}")

print("after training, train questions:", evaluate(encoder_scores, train_q))
print("after training, test questions :", evaluate(encoder_scores, test_q))
```

### البحث

```python
with torch.no_grad():
    page_vecs = encoder(texts)                       # computed once, stored

def search(question, k=3):
    with torch.no_grad():
        scores = page_vecs @ encoder([question])[0]
    top = torch.topk(scores, k)
    return [(ids[i], round(s, 2)) for s, i in zip(top.values.tolist(), top.indices.tolist())]

for question in ["أستخدم كرسياً متحركاً، هل أستطيع الدخول؟", "طفلي عنده حرارة، من يعالجه؟", "Do you speak English?"]:
    print(f"{question}\n   words: {ids[word_scores(question).argmax()]:12} chars: {ids[char_scores(question).argmax()]:12} encoder: {search(question)}")
```

## البحث في ملايين المتجهات

</div>

```python
from sklearn.cluster import KMeans

rng = np.random.default_rng(0)
N, dim, k = 200_000, 64, 10
topics = rng.normal(size=(1000, dim))
X = topics[rng.integers(1000, size=N)] + 0.6 * rng.normal(size=(N, dim))
X = (X / np.linalg.norm(X, axis=1, keepdims=True)).astype(np.float32)
queries = X[rng.choice(N, 200, replace=False)] + 0.05 * rng.normal(size=(200, dim)).astype(np.float32)
queries /= np.linalg.norm(queries, axis=1, keepdims=True)

exact = np.argpartition(-(queries @ X.T), k, axis=1)[:, :k]      # brute force: the true top 10

kmeans = KMeans(n_clusters=256, n_init=1, max_iter=20, random_state=0).fit(X[:50_000])
group = kmeans.predict(X)
members = [np.where(group == g)[0] for g in range(256)]
centers = kmeans.cluster_centers_

for n_groups in (1, 4, 16):
    found = compared = 0
    for q, truth in zip(queries, exact):
        nearest = np.argsort(-(centers @ q))[:n_groups]              # the closest group centers
        cand = np.concatenate([members[g] for g in nearest])
        top = cand[np.argpartition(-(X[cand] @ q), k)[:k]]
        found += len(set(top) & set(truth))
        compared += len(cand)
    print(f"groups searched {n_groups:2}: found {found / (k * len(queries)):.1%} of the true top 10, "
          f"compared with {compared / len(queries) / N:.2%} of the vectors")
```
