## السيغمويد: درجة واحدة إلى احتمال

</div>

```python
import numpy as np
import matplotlib.pyplot as plt

z = np.linspace(-7, 7, 300)
plt.figure(figsize=(6, 3.2))
plt.plot(z, 1 / (1 + np.exp(-z)), lw=2.5)
plt.axhline(0.5, color="gray", ls=":", lw=1)
plt.axvline(0, color="gray", ls=":", lw=1)
plt.xlabel("z (logit)"); plt.ylabel("sigmoid(z)")
plt.grid(True)
plt.show()
```

## بالكود

</div>

```python
import numpy as np

def sigmoid(z):
    return 1 / (1 + np.exp(-z))

def softmax_naive(z):
    e = np.exp(z)
    return e / e.sum()

def softmax(z, T=1.0):
    z = np.asarray(z, dtype=float) / T
    e = np.exp(z - z.max())           # الحيلة: اطرح الأكبر
    return e / e.sum()

print("sigmoid:", sigmoid(np.array([-2, 0, 2])).round(3))
print("softmax:", softmax([2, 1, 0.1]).round(3))
print("T = 0.5:", softmax([2, 1, 0.1], T=0.5).round(3))
print("T = 2  :", softmax([2, 1, 0.1], T=2).round(3))
```

<div dir="rtl">

### الانفجار، بعينك

</div>

```python
big = np.array([1000.0, 999.0])
with np.errstate(over="ignore", invalid="ignore"):
    print("naive :", softmax_naive(big))
print("stable:", softmax(big).round(4))
```

### التلاشي: لماذا نجمع اللوغاريتمات

</div>

```python
probs = np.full(1000, 0.1)          # ألف احتمال، كل واحد 0.1
print("product      :", np.prod(probs))
print("sum of logs  :", round(np.sum(np.log(probs)), 2))
```

```python
def numeric_derivative(f, z, h=1e-5):
    return (f(z + h) - f(z - h)) / (2 * h)

for z in [0.0, 2.0, 5.0]:
    rule = sigmoid(z) * (1 - sigmoid(z))
    print(f"z = {z}: rule = {rule:.5f}, numeric = {numeric_derivative(sigmoid, z):.5f}")
```
