### التقطيع، وأثره

</div>

```python
import numpy as np

handbook = "\n\n".join(f"## {title}\n{body}" for _, title, body in docs)
print(len(handbook.split()), "words in the handbook")

def windows(text, size, overlap):
    words = text.replace("## ", "").split()
    step = size - overlap
    out = []
    for start in range(0, len(words), step):
        out.append(" ".join(words[start:start + size]))
        if start + size >= len(words):
            break
    return out

def by_section(text):
    out = []
    for part in text.split("## ")[1:]:
        title, body = part.strip().split("\n", 1)
        out.append(f"{title}. {body.strip()}")
    return out

strategies = {"windows 40, no overlap": windows(handbook, 40, 0),
              "windows 40, overlap 10": windows(handbook, 40, 10),
              "windows 80, overlap 20": windows(handbook, 80, 20),
              "sections": by_section(handbook)}
for name, chunks in strategies.items():
    cut = [answers[q] for q, _ in test_q if not any(answers[q] in c for c in chunks)]
    print(f"{name:24} {len(chunks):3} chunks | answers cut in two: {cut}")
```

### مقارنة عادلة

</div>

```python
from sklearn.feature_extraction.text import TfidfVectorizer

def char_retriever(chunks):
    vec = TfidfVectorizer(analyzer="char_wb", ngram_range=(2, 4))
    m = vec.fit_transform(chunks)
    return lambda q: (vec.transform([q]) @ m.T).toarray()[0]

def found_within(chunks, budget):
    score = char_retriever(chunks)
    hits = 0
    for q, _ in test_q:
        used = 0
        for i in np.argsort(-score(q), kind="stable"):            # best chunk first
            n = len(chunks[i].split())
            if used > 0 and used + n > budget:
                break                                             # the next chunk does not fit
            used += n
            if answers[q] in chunks[i]:
                hits += 1
                break
    return hits / len(test_q)

print(f"{'':24} {'avg words':>9} | answer found within a budget of 40 / 80 / 120 words")
for name, chunks in strategies.items():
    avg = np.mean([len(c.split()) for c in chunks])
    print(f"{name:24} {avg:9.0f} | " + " / ".join(f"{found_within(chunks, b):.2f}" for b in (40, 80, 120)))
```

### بي إم 25 من الصفر

</div>

```python
import math
import re
from collections import Counter

def normalize(text):
    text = re.sub(r"[ً-ْـ]", "", text)      # diacritics and tatweel
    text = re.sub("[إأآ]", "ا", text)
    return text.replace("ة", "ه").replace("ى", "ي").lower()

def light_stem(w):
    for p in ("وال", "بال", "كال", "فال", "لل", "ال"):
        if w.startswith(p) and len(w) - len(p) >= 2:
            w = w[len(p):]
            break
    for _ in range(2):                                       # at most two suffixes
        for s in ("ات", "ون", "ين", "ها", "هم", "كم", "ني", "ه", "ي", "ك", "ا"):
            if w.endswith(s) and len(w) - len(s) >= 3:
                w = w[:-len(s)]
                break
    return w

def terms(text):
    return [light_stem(w) for w in re.findall(r"\w+", normalize(text))]

class BM25:
    def __init__(self, texts, k1=1.5, b=0.75):
        self.docs = [Counter(terms(t)) for t in texts]
        self.lens = [sum(d.values()) for d in self.docs]
        self.avg = sum(self.lens) / len(self.lens)
        self.k1, self.b = k1, b
        N = len(texts)
        df = Counter(w for d in self.docs for w in d)          # in how many documents is each word
        self.idf = {w: math.log((N - n + 0.5) / (n + 0.5) + 1) for w, n in df.items()}

    def scores(self, query):
        q = terms(query)
        out = np.zeros(len(self.docs))
        for i, (tf, L) in enumerate(zip(self.docs, self.lens)):
            for w in q:
                if w in tf:
                    f = tf[w]
                    out[i] += self.idf[w] * f * (self.k1 + 1) / (f + self.k1 * (1 - self.b + self.b * L / self.avg))
        return out

sections = by_section(handbook)
bm25 = BM25(sections)
print(terms("طلب مني الطبيب رنيناً مغناطيسياً"))
print(terms("الرنين المغناطيسي غير متوفر"))
```

### ثلاث طرق، ثم الدمج

</div>

```python
import zlib
import torch
import torch.nn as nn
import torch.nn.functional as F

def evaluate(score_fn, questions, ks=(1, 3)):
    ranks = []
    for question, gold in questions:
        order = np.argsort(-score_fn(question), kind="stable")
        ranks.append(int(np.where(order == index[gold])[0][0]) + 1)
    ranks = np.array(ranks)
    return {f"recall@{k}": round(float((ranks <= k).mean()), 2) for k in ks}

BUCKETS = 2 ** 15

def pieces(text):
    out = []
    for w in re.findall(r"\w+", text.lower()):
        w = f"<{w}>"
        for n in (2, 3, 4):
            out += [w[i:i + n] for i in range(len(w) - n + 1)]
    return out

class TinyEncoder(nn.Module):
    def __init__(self, dim=64):
        super().__init__()
        self.table = nn.EmbeddingBag(BUCKETS, dim, mode="mean")
        nn.init.normal_(self.table.weight, std=0.1)

    def forward(self, batch):
        flat, starts = [], []
        for text in batch:
            starts.append(len(flat))
            flat += [zlib.crc32(p.encode()) % BUCKETS for p in pieces(text)]
        return F.normalize(self.table(torch.tensor(flat), torch.tensor(starts)), dim=-1)

torch.manual_seed(0)
rng = np.random.default_rng(0)
encoder = TinyEncoder()
questions_of = {}
for question, page in train_q:
    questions_of.setdefault(page, []).append(question)
opt = torch.optim.Adam(encoder.parameters(), lr=0.02)
for step in range(201):
    chosen = rng.choice(len(docs), size=16, replace=False)
    Q = encoder([rng.choice(questions_of[ids[j]]) for j in chosen])
    D = encoder([sections[j] for j in chosen])
    loss = F.cross_entropy(Q @ D.T / 0.1, torch.arange(16))
    opt.zero_grad()
    loss.backward()
    opt.step()

with torch.no_grad():
    section_vecs = encoder(sections)

def encoder_scores(question):
    with torch.no_grad():
        return (section_vecs @ encoder([question])[0]).numpy()

print("encoder ready, final training loss", round(loss.item(), 3))
```

```python
char_scores = char_retriever(sections)

def rrf(*score_fns, k=60):
    def fused(question):
        total = np.zeros(len(sections))
        for fn in score_fns:
            order = np.argsort(-fn(question), kind="stable")
            total[order] += 1 / (k + np.arange(1, len(order) + 1))     # rank 1 gets 1/61, rank 2 gets 1/62 ...
        return total
    return fused

hybrid = rrf(bm25.scores, char_scores, encoder_scores)
for name, fn in [("BM25", bm25.scores), ("char pieces", char_scores), ("tiny encoder", encoder_scores),
                 ("BM25 + encoder", rrf(bm25.scores, encoder_scores)), ("all three", hybrid)]:
    print(f"{name:15}", evaluate(fn, test_q))
```

### الأسئلة التي لا جواب لها

</div>

```python
def best_cosine(question):
    return float(encoder_scores(question).max())

answerable = [best_cosine(q) for q, _ in test_q]
no_answer = [best_cosine(q) for q in unanswerable]
for q, s in sorted(zip(unanswerable, no_answer), key=lambda t: t[1]):
    print(f"{s:.2f}  {q}")

THRESHOLD = 0.55
kept = sum(s >= THRESHOLD for s in answerable)
stopped = sum(s < THRESHOLD for s in no_answer)
print(f"\nthreshold {THRESHOLD}: answerable questions kept {kept}/{len(answerable)} | "
      f"questions without an answer stopped {stopped}/{len(no_answer)}")
```

```python
import json
import threading
from http.server import BaseHTTPRequestHandler, HTTPServer
import anthropic

class FakeAPI(BaseHTTPRequestHandler):
    replies = []
    received = []

    def do_POST(self):
        body = json.loads(self.rfile.read(int(self.headers["Content-Length"])))
        FakeAPI.received.append(body)
        status, reply = FakeAPI.replies.pop(0)
        data = json.dumps(reply).encode()
        self.send_response(status)
        self.send_header("Content-Type", "application/json")
        self.send_header("Content-Length", str(len(data)))
        self.end_headers()
        self.wfile.write(data)

    def log_message(self, *args):
        pass

server = HTTPServer(("127.0.0.1", 0), FakeAPI)
threading.Thread(target=server.serve_forever, daemon=True).start()

def fake_text_reply(text, n_in, n_out):
    return 200, {"id": "msg_fake", "type": "message", "role": "assistant", "model": "fake",
                 "content": [{"type": "text", "text": text}], "stop_reason": "end_turn",
                 "stop_sequence": None, "usage": {"input_tokens": n_in, "output_tokens": n_out}}

client = anthropic.Anthropic(api_key="fake-key", base_url=f"http://127.0.0.1:{server.server_port}", max_retries=0)
```

```python
FALLBACK = "لا أعرف جواب هذا السؤال. اتصل بالعيادة على الرقم 4455."

def check_answer(answer, hits):
    if answer.strip().startswith("لا أعرف"):
        return []                                                  # an honest refusal is fine
    problems = []
    cited = {int(n) for n in re.findall(r"\[(\d+)\]", answer)}
    if not cited:
        problems.append("no citation")
    missing = cited - set(range(1, len(hits) + 1))
    if missing:
        problems.append(f"cites sources that were not given: {sorted(missing)}")
    allowed = set(re.findall(r"\d+", " ".join(hits[n - 1] for n in cited - missing)))   # whole numbers: 30 is not 300
    for number in re.findall(r"\d+", re.sub(r"\[\d+\]", "", answer)):
        if number not in allowed:
            problems.append(f"number {number} is not in the cited sources")
    return problems

def rag(question, k=3):
    if best_cosine(question) < THRESHOLD:                          # first line of defense: nothing relevant
        return {"answer": FALLBACK, "called_model": False}
    top = np.argsort(-hybrid(question))[:k]
    hits = [sections[i] for i in top]
    reply = client.messages.create(model="claude-haiku-4-5-20251001", max_tokens=300,
                                   system=SYSTEM, messages=build_messages(question, hits))
    answer = reply.content[0].text
    return {"answer": answer, "sources": [ids[i] for i in top], "problems": check_answer(answer, hits), "called_model": True}

def show(result):
    print("model called:", result["called_model"], "|", result["answer"])
    if result["called_model"]:
        print("   sources:", result["sources"], "| problems:", result["problems"] or "none")

q = "كم أدفع مقابل زيارة الطبيب العام؟"
FakeAPI.replies.append(fake_text_reply("المعاينة عند الطبيب العام 400 ليرة [1].", 480, 20))
show(rag(q))
FakeAPI.replies.append(fake_text_reply("المعاينة عند الطبيب العام 450 ليرة [1].", 480, 20))    # a hallucinated number
show(rag(q))
FakeAPI.replies.append(fake_text_reply("المعاينة 400 ليرة [4].", 480, 12))                 # a source that does not exist
show(rag(q))
show(rag("هل تبيعون النظارات الطبية؟"))
FakeAPI.replies.append(fake_text_reply("لا أعرف. اتصل بالعيادة على الرقم 4455 لتسأل عن ذلك.", 470, 16))
show(rag("هل يوجد مطعم داخل المبنى؟"))
print("\nrequests that reached the model:", len(FakeAPI.received))
```
