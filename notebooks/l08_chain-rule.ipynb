### مرور أمامي وخلفي لمثال واحد

</div>

```python
def forward_backward(w, b, x, y):
    # المرور الأمامي
    y_hat = w * x + b
    e = y_hat - y
    L = e ** 2
    # المرور الخلفي: من اليمين إلى اليسار
    dL_de = 2 * e
    dL_dyhat = dL_de * 1
    dL_db = dL_dyhat * 1
    dL_dw = dL_dyhat * x
    return L, dL_dw, dL_db

L, dw, db = forward_backward(w=1, b=1, x=2, y=7)
print("loss:", L, " dL/dw:", dw, " dL/db:", db)
```

### نتحقق عددياً

</div>

```python
def loss(w, b, x=2, y=7):
    return (w * x + b - y) ** 2

h = 1e-5
num_dw = (loss(1 + h, 1) - loss(1 - h, 1)) / (2 * h)
num_db = (loss(1, 1 + h) - loss(1, 1 - h)) / (2 * h)
print("numeric dL/dw:", round(num_dw, 4), " numeric dL/db:", round(num_db, 4))
```

### التدرّج على بيانات كاملة

</div>

```python
import numpy as np

x = np.array([1, 2, 3])
y = np.array([2, 4, 5])

def mse(w, b):
    return np.mean((w * x + b - y) ** 2)

def mse_grad(w, b):
    err = w * x + b - y                 # الخطأ لكل مثال
    dw = 2 * np.mean(err * x)           # الخطأ مضروباً في المدخل
    db = 2 * np.mean(err)
    return dw, db

w, b = 0.0, 0.0
dw, db = mse_grad(w, b)
num_dw = (mse(w + h, b) - mse(w - h, b)) / (2 * h)
num_db = (mse(w, b + h) - mse(w, b - h)) / (2 * h)
print(f"formula: dw = {dw:.4f}, db = {db:.4f}")
print(f"numeric: dw = {num_dw:.4f}, db = {num_db:.4f}")
```
