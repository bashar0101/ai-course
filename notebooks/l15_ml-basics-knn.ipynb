### من الصفر

</div>

```python
import numpy as np

X_train = np.array([[1, 0], [4, 0], [3, 1], [5, 5], [5, 2], [6, 3]])
y_train = np.array([0, 0, 0, 1, 1, 1])      # 0 رسب، 1 نجح

def knn_predict(x_new, k):
    distances = np.linalg.norm(X_train - x_new, axis=1)   # مسافة لكل صف
    nearest = np.argsort(distances)[:k]                   # أرقام أقرب k
    votes = y_train[nearest]
    return np.bincount(votes).argmax()                    # الفئة الأكثر

for k in [1, 3, 5]:
    print(f"k = {k}: prediction = {knn_predict(np.array([3, 3]), k)}")
```

<div dir="rtl">

### بمكتبة سايكت ليرن

</div>

```python
from sklearn.neighbors import KNeighborsClassifier

model = KNeighborsClassifier(n_neighbors=3)
model.fit(X_train, y_train)
print("prediction:", model.predict([[3, 3]]))
print("probabilities:", model.predict_proba([[3, 3]]).round(3))
```

### بيانات حقيقية: أنواع النبيذ

</div>

```python
from sklearn.datasets import load_wine
from sklearn.model_selection import train_test_split

X, y = load_wine(return_X_y=True)
print("shape:", X.shape, " classes:", np.bincount(y))
print("smallest column max:", X.max(axis=0).min(), " largest column max:", X.max(axis=0).max())

X_tr, X_te, y_tr, y_te = train_test_split(X, y, test_size=0.3, random_state=0, stratify=y)
print("train:", X_tr.shape, " test:", X_te.shape)
```

```python
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler

raw = KNeighborsClassifier(n_neighbors=5).fit(X_tr, y_tr)
scaled = make_pipeline(StandardScaler(), KNeighborsClassifier(n_neighbors=5)).fit(X_tr, y_tr)

print(f"accuracy without scaling: {raw.score(X_te, y_te):.3f}")
print(f"accuracy with scaling   : {scaled.score(X_te, y_te):.3f}")
```

<div dir="rtl">

### أثر عدد الجيران

</div>

```python
import matplotlib.pyplot as plt

ks = range(1, 41)
train_acc, test_acc = [], []
for k in ks:
    m = make_pipeline(StandardScaler(), KNeighborsClassifier(n_neighbors=k)).fit(X_tr, y_tr)
    train_acc.append(m.score(X_tr, y_tr))
    test_acc.append(m.score(X_te, y_te))

plt.figure(figsize=(6, 3.4))
plt.plot(ks, train_acc, lw=2, label="train accuracy")
plt.plot(ks, test_acc, lw=2, label="test accuracy")
plt.xlabel("k (number of neighbors)"); plt.ylabel("accuracy")
plt.legend(); plt.grid(True)
plt.show()
```
