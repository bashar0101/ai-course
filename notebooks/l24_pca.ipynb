### بنمباي، على النقاط الخمس

</div>

```python
import numpy as np

P = np.array([[-2, -1], [-1, -2], [0, 0], [1, 2], [2, 1]], dtype=float)
Pc = P - P.mean(axis=0)                  # المتوسط صفر هنا أصلاً
cov = Pc.T @ Pc / len(Pc)
values, vectors = np.linalg.eigh(cov)    # للمصفوفات المتناظرة
order = np.argsort(values)[::-1]         # من الأكبر
values, vectors = values[order], vectors[:, order]

print("covariance:\n", cov)
print("eigenvalues (variances):", values.round(3))
print("first component:", vectors[:, 0].round(3))
print("explained ratio:", (values / values.sum()).round(3))
print("projections on PC1:", (Pc @ vectors[:, 0]).round(3))
```

<div dir="rtl">

### بمكتبة سايكت ليرن

</div>

```python
from sklearn.decomposition import PCA

pca = PCA(n_components=2).fit(P)
print("components:\n", pca.components_.round(3))
print("explained variance ratio:", pca.explained_variance_ratio_.round(3))
```

<div dir="rtl">

### من 64 بعداً إلى بعدين: صور الأرقام

</div>

```python
import matplotlib.pyplot as plt
from sklearn.datasets import load_digits
from sklearn.preprocessing import StandardScaler

digits = load_digits()
X, y = digits.data, digits.target
print("shape:", X.shape)

Z = PCA(n_components=2).fit_transform(StandardScaler().fit_transform(X))

plt.figure(figsize=(6, 4.6))
sc = plt.scatter(Z[:, 0], Z[:, 1], c=y, cmap="tab10", s=8)
plt.colorbar(sc, ticks=range(10), label="digit")
plt.xlabel("PC 1"); plt.ylabel("PC 2")
plt.show()
```

<div dir="rtl">

### كم مكوّناً نحتاج؟

</div>

```python
pca_full = PCA().fit(StandardScaler().fit_transform(X))
cum = np.cumsum(pca_full.explained_variance_ratio_)

plt.figure(figsize=(6, 3.2))
plt.plot(range(1, len(cum) + 1), cum, lw=2)
plt.axhline(0.9, color="C1", ls="--", label="90%")
plt.xlabel("number of components"); plt.ylabel("cumulative explained variance")
plt.legend(); plt.grid(True)
plt.show()

print("components for 90% of the variance:", int(np.argmax(cum >= 0.9)) + 1)
```
