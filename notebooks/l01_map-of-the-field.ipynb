## أول تجربة بالكود

</div>

```python
from sklearn.linear_model import LinearRegression

X = [[50], [80], [100]]    # المدخلات: المساحة
y = [100, 160, 200]        # الأجوبة الصحيحة: السعر

model = LinearRegression()
model.fit(X, y)            # التدريب: البحث عن القاعدة

print("learned slope:", round(model.coef_[0], 3))
print("prediction for 70 m2:", round(model.predict([[70]])[0], 1))
```
