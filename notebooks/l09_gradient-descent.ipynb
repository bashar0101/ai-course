### بالكود

</div>

```python
import numpy as np

x = np.array([1, 2, 3])
y = np.array([2, 4, 5])

def mse(w, b):
    return np.mean((w * x + b - y) ** 2)

def grad(w, b):
    err = w * x + b - y
    return 2 * np.mean(err * x), 2 * np.mean(err)

w, b = 0.0, 0.0
lr = 0.1
history = []
for step in range(300):
    dw, db = grad(w, b)
    w -= lr * dw               # الخوارزمية كلها في هذين السطرين
    b -= lr * db
    history.append(mse(w, b))
    if step in (0, 1, 2, 10, 50, 299):
        print(f"step {step:3d}: w = {w:.4f}, b = {b:.4f}, loss = {history[-1]:.5f}")
```

<div dir="rtl">

### منحنى الخسارة

</div>

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(6, 3.4))
plt.plot(history, lw=2)
plt.yscale("log")
plt.xlabel("step"); plt.ylabel("loss (log scale)")
plt.grid(True)
plt.show()
```

<div dir="rtl">

### نقارن معدلات تعلّم مختلفة

</div>

```python
def train(lr, steps=30):
    w, b = 0.0, 0.0
    losses = []
    for _ in range(steps):
        dw, db = grad(w, b)
        w, b = w - lr * dw, b - lr * db
        losses.append(mse(w, b))
    return losses

plt.figure(figsize=(6, 3.4))
for lr in [0.01, 0.1, 0.19]:
    plt.plot(train(lr), lw=2, label=f"lr = {lr}")
plt.yscale("log")
plt.xlabel("step"); plt.ylabel("loss (log scale)")
plt.grid(True); plt.legend()
plt.show()

print("loss after 30 steps with lr=0.19:", f"{train(0.19)[-1]:.1f}")
```
