### من الصفر

</div>

```python
import numpy as np

def kmeans(X, init_centers, max_iter=100):
    C = np.array(init_centers, dtype=float)
    labels = None
    for it in range(max_iter):
        d = ((X[:, None, :] - C[None, :, :]) ** 2).sum(axis=2)   # مسافة كل نقطة إلى كل مركز
        new_labels = d.argmin(axis=1)                             # الإسناد
        if labels is not None and np.array_equal(new_labels, labels):
            break
        labels = new_labels
        C = np.array([X[labels == k].mean(axis=0) for k in range(len(C))])   # التحديث
    inertia = ((X - C[labels]) ** 2).sum()
    return C, labels, inertia, it

P = np.array([[1, 1], [1, 2], [2, 1], [6, 5], [7, 6], [6, 6]])
C, labels, inertia, it = kmeans(P, init_centers=[[1, 1], [1, 2]])
print("centers:\n", C.round(3))
print("labels:", labels, " inertia:", round(inertia, 3), " stopped at round", it)
```

<div dir="rtl">

### بمكتبة سايكت ليرن، على بيانات أكبر

</div>

```python
from sklearn.cluster import KMeans
from sklearn.datasets import make_blobs

X, true_groups = make_blobs(n_samples=400, centers=4, cluster_std=1.0, random_state=7)
km = KMeans(n_clusters=4, n_init=10, random_state=0).fit(X)
print("centers:\n", km.cluster_centers_.round(2))
print("inertia:", round(km.inertia_, 1))
print("sizes:", np.bincount(km.labels_))
```

<div dir="rtl">

### المرفق والصورة الظلية

</div>

```python
import matplotlib.pyplot as plt
from sklearn.metrics import silhouette_score

ks = range(1, 9)
inertias, sils = [], []
for k in ks:
    m = KMeans(n_clusters=k, n_init=10, random_state=0).fit(X)
    inertias.append(m.inertia_)
    sils.append(silhouette_score(X, m.labels_) if k > 1 else np.nan)

fig, (a1, a2) = plt.subplots(1, 2, figsize=(9, 3.2))
a1.plot(ks, inertias, "o-", lw=2); a1.set_xlabel("K"); a1.set_ylabel("inertia"); a1.grid(True)
a2.plot(ks, sils, "o-", lw=2, color="C1"); a2.set_xlabel("K"); a2.set_ylabel("silhouette"); a2.grid(True)
plt.tight_layout()
plt.show()

print("best K by silhouette:", ks[int(np.nanargmax(sils))])
```

### متى تفشل كي-مينز؟

</div>

```python
from sklearn.datasets import make_moons

Xm, _ = make_moons(n_samples=300, noise=0.06, random_state=0)
lab = KMeans(n_clusters=2, n_init=10, random_state=0).fit_predict(Xm)

plt.figure(figsize=(5.5, 3.2))
plt.scatter(Xm[:, 0], Xm[:, 1], c=lab, cmap="coolwarm", s=14)
plt.title("k-means on two moons")
plt.show()
```
