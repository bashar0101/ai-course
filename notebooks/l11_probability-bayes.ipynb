### نحاكي مليون شخص

</div>

```python
import numpy as np

rng = np.random.default_rng(0)
n = 1_000_000

sick = rng.random(n) < 0.01                       # 1% مرضى
positive = np.where(sick,
                    rng.random(n) < 0.90,         # المريض: إيجابي 90%
                    rng.random(n) < 0.09)         # السليم: إيجابي 9%

print("people with a positive test:", positive.sum())
print("P(sick | positive) =", round(sick[positive].mean(), 4))
```

### دالة بايز

</div>

```python
def bayes(prior, p_pos_if_sick, p_pos_if_healthy):
    evidence = p_pos_if_sick * prior + p_pos_if_healthy * (1 - prior)
    return p_pos_if_sick * prior / evidence

print("rare disease (1%)  :", round(bayes(0.01, 0.90, 0.09), 4))
print("common disease (20%):", round(bayes(0.20, 0.90, 0.09), 4))
print("second test positive:", round(bayes(bayes(0.01, 0.90, 0.09), 0.90, 0.09), 4))
```
