### خطي، والمعامل

</div>

```python
import numpy as np
from sklearn.svm import SVC
from sklearn.datasets import make_blobs

X, y = make_blobs(n_samples=80, centers=[[-1.5, -1], [1.5, 1]], cluster_std=1.1, random_state=4)

for C in [0.01, 1, 100]:
    m = SVC(kernel="linear", C=C).fit(X, y)
    width = 2 / np.linalg.norm(m.coef_)
    print(f"C = {C:<5} support vectors = {len(m.support_):2d}   street width = {width:.2f}   train acc = {m.score(X, y):.3f}")
```

```python
import matplotlib.pyplot as plt

m = SVC(kernel="linear", C=1).fit(X, y)
xx, yy = np.meshgrid(np.linspace(-5, 5, 300), np.linspace(-5, 5, 300))
Z = m.decision_function(np.column_stack([xx.ravel(), yy.ravel()])).reshape(xx.shape)

plt.figure(figsize=(5.2, 4.4))
plt.contour(xx, yy, Z, levels=[-1, 0, 1], colors=["gray", "k", "gray"], linestyles=["--", "-", "--"])
plt.scatter(X[y == 0, 0], X[y == 0, 1], s=18, color="C0")
plt.scatter(X[y == 1, 0], X[y == 1, 1], s=18, color="C1", marker="s")
sv = m.support_vectors_
plt.scatter(sv[:, 0], sv[:, 1], s=110, facecolors="none", edgecolors="C3", linewidths=1.5, label="support vectors")
plt.legend(loc="upper left")
plt.show()
```

### حين لا يكفي الخط: القمران

</div>

```python
from sklearn.datasets import make_moons
from sklearn.model_selection import train_test_split
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler

Xm, ym = make_moons(n_samples=300, noise=0.2, random_state=0)
X_tr, X_te, y_tr, y_te = train_test_split(Xm, ym, test_size=0.3, random_state=0)

for kernel, gamma in [("linear", "scale"), ("rbf", "scale"), ("rbf", 100)]:
    model = make_pipeline(StandardScaler(), SVC(kernel=kernel, gamma=gamma)).fit(X_tr, y_tr)
    print(f"{kernel:6} gamma={str(gamma):5}  train = {model.score(X_tr, y_tr):.3f}   test = {model.score(X_te, y_te):.3f}")
```
