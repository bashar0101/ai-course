### انتبه: من هي الفئة الموجبة؟

</div>

```python
import numpy as np
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression

X, y_raw = load_breast_cancer(return_X_y=True)
y = 1 - y_raw                     # الآن: 1 خبيث (موجب)، 0 حميد
print("malignant share:", round(y.mean(), 3))

X_tr, X_te, y_tr, y_te = train_test_split(X, y, test_size=0.3, random_state=0, stratify=y)
model = make_pipeline(StandardScaler(), LogisticRegression(max_iter=1000)).fit(X_tr, y_tr)
y_pred = model.predict(X_te)
```

<div dir="rtl">

### المصفوفة والمقاييس

</div>

```python
from sklearn.metrics import confusion_matrix, accuracy_score, precision_score, recall_score, f1_score

cm = confusion_matrix(y_te, y_pred)
print(cm)
tn, fp, fn, tp = cm.ravel()
print(f"TP={tp} FN={fn} FP={fp} TN={tn}")
print("accuracy :", round(accuracy_score(y_te, y_pred), 3))
print("precision:", round(precision_score(y_te, y_pred), 3))
print("recall   :", round(recall_score(y_te, y_pred), 3))
print("f1       :", round(f1_score(y_te, y_pred), 3))
```

```python
import matplotlib.pyplot as plt
from sklearn.metrics import ConfusionMatrixDisplay

ConfusionMatrixDisplay.from_predictions(y_te, y_pred, display_labels=["benign", "malignant"], cmap="Blues", colorbar=False)
plt.show()
```

<div dir="rtl">

### تقرير كامل في سطر

</div>

```python
from sklearn.metrics import classification_report
print(classification_report(y_te, y_pred, target_names=["benign", "malignant"], digits=3))
```

<div dir="rtl">

### فئات كثيرة: أنواع النبيذ

</div>

```python
from sklearn.datasets import load_wine

Xw, yw = load_wine(return_X_y=True)
Xw_tr, Xw_te, yw_tr, yw_te = train_test_split(Xw, yw, test_size=0.3, random_state=0, stratify=yw)
from sklearn.neighbors import KNeighborsClassifier

mw = KNeighborsClassifier(n_neighbors=5).fit(Xw_tr, yw_tr)     # ضعيف عن قصد: بدون توحيد
pw = mw.predict(Xw_te)
print(confusion_matrix(yw_te, pw))
print("macro F1   :", round(f1_score(yw_te, pw, average="macro"), 3))
print("weighted F1:", round(f1_score(yw_te, pw, average="weighted"), 3))
print(classification_report(yw_te, pw, digits=3))
```

```python
y_fraud = np.array([1] * 10 + [0] * 990)
lazy = np.zeros(1000, dtype=int)
print("accuracy:", accuracy_score(y_fraud, lazy))
print("recall  :", recall_score(y_fraud, lazy))
print("f1      :", f1_score(y_fraud, lazy, zero_division=0))
```
