### طرق سريعة لصنع المصفوفات

</div>

```python
print(np.zeros(3))
print(np.ones((2, 2)))
print(np.arange(0, 10, 2))        # من 0 إلى ما قبل 10، بخطوة 2
print(np.linspace(0, 1, 5))       # خمسة أرقام متساوية المسافة من 0 إلى 1

rng = np.random.default_rng(42)   # مولد أرقام عشوائية ببذرة ثابتة
print(rng.integers(1, 7, size=5)) # خمس رميات نرد
```

### الوصول إلى العناصر

</div>

```python
m = np.array([[1, 2, 3],
              [4, 5, 6]])

print(m[1, 2])     # الصف 1، العمود 2
print(m[0])        # الصف الأول كله
print(m[:, 1])     # العمود 1 من كل الصفوف
print(m[m > 3])    # العناصر الأكبر من 3
```

### العمليات على المصفوفة كلها

</div>

```python
x = np.array([1, 2, 3, 4])
print(x * 2)
print(x ** 2)
print(x + np.array([10, 20, 30, 40]))
print(x.sum(), x.mean(), x.max(), x.argmax())
```

```python
import time

data = np.arange(1_000_000)

t = time.perf_counter()
total = 0
for v in data:
    total += v * v
loop_time = time.perf_counter() - t

t = time.perf_counter()
total2 = np.sum(data * data)
np_time = time.perf_counter() - t

print("same result:", total == total2)
print(f"numpy is about {loop_time / np_time:.0f} times faster")
```
