## بالكود

```python
import numpy as np

delays = np.array([2, 4, 4, 4, 5, 5, 7, 9])
print("mean:", delays.mean(), " median:", np.median(delays))
print("variance:", delays.var(), " std:", delays.std())
print("std with n-1:", round(delays.std(ddof=1), 3))

with_outlier = delays.copy()
with_outlier[-1] = 45
print("with outlier -> mean:", with_outlier.mean(), " median:", np.median(with_outlier))
```

<div dir="rtl">

### التوحيد بيدك، وبمكتبة سايكت ليرن

</div>

```python
z = (delays - delays.mean()) / delays.std()
mm = (delays - delays.min()) / (delays.max() - delays.min())
print("z-scores:", z)
print("min-max :", mm.round(3))

from sklearn.preprocessing import StandardScaler, MinMaxScaler

col = delays.reshape(-1, 1)          # المكتبة تنتظر جدولاً: صف لكل مثال
print("sklearn z:", StandardScaler().fit_transform(col).ravel())
```

### التوزيع الملتوي

</div>

```python
import matplotlib.pyplot as plt

rng = np.random.default_rng(1)
salaries = rng.lognormal(mean=3.4, sigma=0.5, size=2000)   # بيانات مصنوعة

plt.figure(figsize=(6, 3.4))
plt.hist(salaries, bins=60, color="C0", alpha=0.75)
plt.axvline(salaries.mean(), color="C1", lw=2.5, label=f"mean = {salaries.mean():.1f}")
plt.axvline(np.median(salaries), color="C2", lw=2.5, ls="--", label=f"median = {np.median(salaries):.1f}")
plt.xlabel("salary (thousands)"); plt.ylabel("count")
plt.legend()
plt.show()
```
