## بالكود

</div>

```python
xs = [1, 2, 3]   # المدخلات
ys = [2, 4, 5]   # الأجوبة الصحيحة

def predict(x, w, b):
    return w * x + b

def mse(w, b):
    total = 0
    for x, y in zip(xs, ys):
        error = predict(x, w, b) - y
        total += error ** 2
    return total / len(xs)

print("model A:", round(mse(2, 0), 3))
print("model B:", round(mse(1.5, 1), 3))
```

```python
import numpy as np

x = np.array([1, 2, 3])
y = np.array([2, 4, 5])

def mse(w, b):
    y_hat = w * x + b              # كل التوقعات دفعة واحدة
    return np.mean((y_hat - y) ** 2)

print("predictions of B:", 1.5 * x + 1)
print("MSE of B:", round(mse(1.5, 1), 4))
```

### التعلّم كبحث: نجرب آلاف الخطوط

</div>

```python
best = None
for w in np.arange(0, 3.01, 0.05):
    for b in np.arange(-2, 3.01, 0.05):
        loss = mse(w, b)
        if best is None or loss < best[0]:
            best = (loss, w, b)

loss, w, b = best
print(f"best w = {w:.2f}, best b = {b:.2f}, MSE = {loss:.4f}")
```

```python
ws = np.arange(0, 3.01, 0.05)
bs = np.arange(-2, 3.01, 0.05)
print("tries in our search:", len(ws) * len(bs))

# لو جربنا 100 قيمة لكل معامل:
print("2 parameters :", 100 ** 2, "tries")
print("10 parameters:", f"{100 ** 10:.0e}", "tries")

seconds = 100 ** 10 / 1e8        # جهاز يجرب مئة مليون خط في الثانية
print("years:", round(seconds / (3600 * 24 * 365)))
```
