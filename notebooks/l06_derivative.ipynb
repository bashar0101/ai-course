## بالكود: المشتقة العددية

</div>

```python
def f(w):
    return (w - 3) ** 2

def f_prime(w):          # من القاعدة
    return 2 * w - 6

def numeric_derivative(func, w, h=1e-5):
    return (func(w + h) - func(w - h)) / (2 * h)

for w in [5, 1, 3, 3.5]:
    print(f"w = {w:>3}:  rule = {f_prime(w):6.3f}   numeric = {numeric_derivative(f, w):6.3f}")
```

<div dir="rtl">

### ميل الوتر يقترب من المشتقة

</div>

```python
for h in [1, 0.1, 0.01, 0.001]:
    secant = (f(5 + h) - f(5)) / h
    print(f"h = {h:<6} secant slope = {secant:.4f}")
```

### نرسم المنحنى والمماس

</div>

```python
import numpy as np
import matplotlib.pyplot as plt

w = np.linspace(-1, 7, 200)
w0 = 5
tangent = f(w0) + f_prime(w0) * (w - w0)

plt.figure(figsize=(6, 3.6))
plt.plot(w, f(w), lw=2.5, label="f(w) = (w - 3)^2")
plt.plot(w, tangent, "--", lw=2, label="tangent at w = 5, slope 4")
plt.scatter([w0], [f(w0)], s=60, zorder=3, color="C1")
plt.ylim(-2, 16)
plt.xlabel("w"); plt.ylabel("f(w)")
plt.grid(True); plt.legend()
plt.show()
```
